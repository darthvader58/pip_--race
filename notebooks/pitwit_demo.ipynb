# PitWit Library Demo

This notebook demonstrates inference, data products, visualization output, benchmarking, and smoke checks.

```python
from pip_race import HpcTelemetryPacket, PitWit, run_pitwit_benchmark
from pip_race.data import frames_to_rows, summarize_frames
from pip_race.visualization import pit_risk_svg, pit_risk_vega_spec
from IPython.display import SVG, display
```

```python
packets = [
    HpcTelemetryPacket(car_id="ALB", lap=42, lap_distance_m=2300, speed_kph=188, tire_age_laps=27, compound="MEDIUM"),
    HpcTelemetryPacket(car_id="ALB", lap=42, lap_distance_m=2410, speed_kph=178, tire_age_laps=29, compound="MEDIUM", track_status="VSC"),
    HpcTelemetryPacket(car_id="SAR", lap=42, lap_distance_m=2390, speed_kph=181, tire_age_laps=18, compound="HARD"),
]

pitwit = PitWit()
frames = pitwit.process_many(packets)
summarize_frames(frames)
```

```python
frames_to_rows(frames)
```

```python
display(SVG(pit_risk_svg(frames)))
```

```python
pit_risk_vega_spec(frames)
```

```python
run_pitwit_benchmark(iterations=100, warmup=10).to_dict()
```

```python
assert len(frames) == 3
assert all(0 <= frame.inference.pit_risk <= 1 for frame in frames)
assert pit_risk_svg(frames).startswith("<svg")
print("PitWit notebook smoke checks passed")
```
